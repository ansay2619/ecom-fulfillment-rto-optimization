```python
import pandas as pd
import random

import pandas as pd

print("🧹 Running Python data-cleaning and product categorization pipeline...")

# Import the raw dataset
df_raw = pd.read_csv("raw_ecommerce_fulfillment_data.csv")

print(f"Raw dataset loaded successfully: {len(df_raw)} records")

# Product categorization function
def categorize_product(title):
    title_lower = title.lower()

    if ('saree' in title_lower or
        'kurti' in title_lower or
        'palazzo' in title_lower or
        'kurta' in title_lower or
        'lehenga' in title_lower):
        return "Ethnic Wear"

    elif ('shoes' in title_lower or
          'sandals' in title_lower or
          'sneakers' in title_lower or
          'loafers' in title_lower or
          'joggers' in title_lower):
        return "Footwear"

    elif ('earphones' in title_lower or
          'cable' in title_lower or
          'watch' in title_lower):
        return "Electronics"

    else:
        return "General Merchandise"


# Create a clean, normalized Product_Category column
df_raw["Product_Category"] = df_raw["Raw_Product_Title"].apply(
    categorize_product
)

# Remove the raw product title after categorization
df_clean = df_raw.drop(columns=["Raw_Product_Title"])

# Keep only the required analytical columns
df_clean = df_clean[
    [
        "Order_ID",
        "Warehouse",
        "Customer_City",
        "Payment_Method",
        "Brand_Name",
        "Product_Category",
        "Order_Value_INR",
        "Hours_to_Dispatch",
        "Order_Status"
    ]
]

# Export the cleaned dataset
df_clean.to_csv(
    "clean_ecommerce_fulfillment_data.csv",
    index=False
)

print("✅ Data cleaning completed successfully!")
print("📁 Clean dataset saved as: clean_ecommerce_fulfillment_data.csv")

# Preview cleaned data
print("\n👀 Cleaned Data Preview:")
print(df_clean.head())

# Check dataset dimensions
print("\n📊 Cleaned Dataset Shape:", df_clean.shape)
```
