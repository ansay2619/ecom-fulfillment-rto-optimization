### 1. Load Data

```python
import pandas as pd
import numpy as np

df = pd.read_csv('raw_data.csv')
df.head()
```

### 2. Identify and Remove Duplicates

```python
# Check duplicate count
display(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()
```

### 3. Explore and Standardize Categories

```python
# Check unique values of payment method
display(df['Payment_Method'].unique())
```

```python
# Standardize Payment Methods
payment_map = {'C.O.D.': 'COD', 'cod': 'COD', 'Cash on Delivery': 'COD', 'Cash': 'COD'}
df['Payment_Method'] = df['Payment_Method'].replace(payment_map)
```

```python
# Check unique values of cities
display(df['Customer_City'].unique())
```

```python
# Standardize Cities
df['Customer_City'] = df['Customer_City'].replace(['mumbai', 'MUM', 'Mumbay'], 'Mumbai')
df['Customer_City'] = df['Customer_City'].replace(['lucknow', 'LKO'], 'Lucknow')
df['Customer_City'] = df['Customer_City'].replace(['bengaluru', 'Bengaluru', 'Bnglr'], 'Bangalore')
df['Customer_City'] = df['Customer_City'].replace(['ranchi'], 'Ranchi')
```

```python
# Clean formatting
df['Warehouse'] = df['Warehouse'].str.strip()
df['Brand_Name'] = df['Brand_Name'].str.title()
```

```python
df
```

### 4. Explore and Fix Numerical Errors

```python
# Check for impossible negative time values
display((df['Hours_to_Dispatch'] < 0).sum())
```

```python
# Fix negatives with absolute values
df['Hours_to_Dispatch'] = df['Hours_to_Dispatch'].abs()
```

```python
# Fix string numbers (e.g. '1,500.00')
df['Order_Value_INR'] = df['Order_Value_INR'].astype(str).str.replace(',', '').astype(float)
```

```python
# Fix negatives with absolute values
df['Order_Value_INR'] = df['Order_Value_INR'].abs()
```

```python
# Cap massive outliers
df.loc[df['Order_Value_INR'] > 100000, 'Order_Value_INR'] = np.nan
```

```python
df
```

### 5. Explore and Impute Missing Values

```python
# Check null counts across all columns
display(df.isnull().sum())
```

```python
# Impute missing categories
df['Customer_City'] = df['Customer_City'].fillna('Unknown')
```

```python
df['Brand_Name'] = df['Brand_Name'].fillna('Unknown')
```

```python
# Impute numerical blanks with median
df['Order_Value_INR'] = df['Order_Value_INR'].fillna(df['Order_Value_INR'].median())
```

```python
df
```

### 6. Save Final Dataset

```python
df.to_csv('cleaned_data.csv', index=False)
df.head()
```
